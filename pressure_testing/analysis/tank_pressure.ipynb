```python
fileloc = 'G:\\Shared drives\WHOI Argo\BGC\Float Tracking\pressure_test_data\\tankdata\\'
filename = 'PT_1557_1558_1559.dat'
filenameout='PT_1557_1558_1559.xlsx'
```

```python
data = pd.read_table(fileloc+filename)
data = data[14:]
data.columns = ['dat']
print(data.columns)
DATA = data['dat'].str.split(',', expand = True)
DATA.columns = ['date', 'pressure', 'd', 'f', 'g']
DATA.drop(['d', 'f', 'g'], axis = 1)
DATA['pressure'].replace('', np.nan, inplace = True)
DATA.dropna(subset = ['pressure'], inplace = True)
DATA

```
